# Exploratory Data Analysis - Day 1
This notebook demonstrates dataset properties and noise profiling for the KLA Image Restoration dataset.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

## 1. Verify Shapes and Ranges

```python
gt_dir = '../data/raw/train/GT'
noisy_dir = '../data/raw/train/NoisyLR'
files = sorted(os.listdir(gt_dir))
print(f'Total training files: {len(files)}')
sample_gt = np.load(os.path.join(gt_dir, files[0]))
sample_noisy = np.load(os.path.join(noisy_dir, files[0]))
print(f'GT shape: {sample_gt.shape}, range: [{sample_gt.min():.4f}, {sample_gt.max():.4f}]')
print(f'Noisy shape: {sample_noisy.shape}, range: [{sample_noisy.min():.4f}, {sample_noisy.max():.4f}]')
```
